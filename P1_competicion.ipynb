```python
import pandas as pd
import joblib

df_competicion = pd.read_pickle("bank_competition.pkl")

df_competicion["contacted"] = (df_competicion["pdays"] != -1).astype(int)
df_competicion["pdays"] = df_competicion["pdays"].replace(-1, 0)

pack = joblib.load('modelo_final.joblib')
modelo_cargado = pack["pipeline"]
prediccion = modelo_cargado.predict(df_competicion)

df_predicciones = pd.DataFrame(prediccion, columns=['deposit'])

df_predicciones.to_csv('predicciones.csv', index=False)
print("Predicciones guardadas correctamente en predicciones.csv")
```
